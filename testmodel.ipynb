```python
import torch
from BPEtokenizer import BPETokenizer
from minGPT import MiniChatGPT
```

```python


# ============================================================
# DEVICE
# ============================================================

device = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)


# ============================================================
# LOAD CHECKPOINT
# ============================================================

checkpoint = torch.load(
    "minigpt_checkpoint.pt",
    map_location=device
)


print(
    "Checkpoint step:",
    checkpoint["step"]
)


# ============================================================
# LOAD CONFIG
# ============================================================

config = checkpoint[
    "config"
]


print(
    "Config:",
    config
)


# ============================================================
# RESTORE TOKENIZER
# ============================================================

tokenizer_config = checkpoint[
    "tokenizer"
]


tokenizer = BPETokenizer(
    num_merges=
        tokenizer_config[
            "num_merges"
        ]
)


tokenizer.merges = (
    tokenizer_config[
        "merges"
    ]
)


tokenizer.token_to_id = (
    tokenizer_config[
        "token_to_id"
    ]
)


tokenizer.id_to_token = (
    tokenizer_config[
        "id_to_token"
    ]
)


# ============================================================
# CREATE MODEL
# ============================================================

model = MiniChatGPT(

    vocal_size_v=
        config[
            "vocab_size"
        ],

    block_size=
        config[
            "block_size"
        ],

    n_emb_C=
        config[
            "n_emb"
        ],

    n_head_H=
        config[
            "n_head"
        ],

    n_layer=
        config[
            "n_layer"
        ]
)


# ============================================================
# LOAD WEIGHTS
# ============================================================

model.load_state_dict(
    checkpoint[
        "model_state_dict"
    ]
)


model = model.to(
    device
)


model.eval()


print(
    "Model loaded successfully."
)


# ============================================================
# TEST PROMPT
# ============================================================

prompt = "Trí tuệ nhân tạo"


# Text -> Token IDs
prompt_ids = tokenizer.encode(
    prompt
)


print(
    "Prompt IDs:",
    prompt_ids
)


# [T]
# ↓
# [1,T]

idx = torch.tensor(
    [prompt_ids],
    dtype=torch.long,
    device=device
)


# ============================================================
# GENERATE
# ============================================================

with torch.no_grad():
    generated = model.generate(
        idx,
        max_new_token=100
    )


# ============================================================
# DECODE
# ============================================================

generated_ids = (
    generated[0]
    .tolist()
)


text = tokenizer.decode(
    generated_ids
)


print(
    "\n=============================="
)

print(
    "PROMPT:"
)

print(
    prompt
)


print(
    "\nMODEL OUTPUT:"
)

print(
    text
)

print(
    "=============================="
)
```
